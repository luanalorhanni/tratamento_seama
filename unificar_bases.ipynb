```python
import pandas as pd
import os
```

```python
#pastas = ['ACERTO_POR_HABILIDADE_2_A_9', 'HABILIDADE_DESEMPENHO_ESCOLA_2_A_9', 'HABILIDADE_DESEMPENHO_MUNICIPIO_1_A_9', 'HABILIDADE_DESEMPENHO_TURMA_1_A_9']
pastas = ['HABILIDADE_DESEMPENHO_TURMA_1_A_9']
```

```python
for pasta in pastas:
    dfs = []

    for arquivo in os.listdir(pasta):
        caminho_arquivo = os.path.join(pasta, arquivo)

        try:
            if arquivo.endswith('.xlsx'):
                df = pd.read_excel(caminho_arquivo)
            elif arquivo.endswith('.csv'):
                df = pd.read_csv(caminho_arquivo, sep=';', engine='python')
            else:
                continue  # pula outros tipos de arquivo

            dfs.append(df)
        except Exception as e:
            print(f'Erro ao ler {arquivo} na pasta {pasta}: {e}')

    if dfs:
        df_unificado = pd.concat(dfs, ignore_index=True)
        nome_saida = f'{pasta}_unificado.xlsx'
        df_unificado.to_excel(nome_saida, index=False)
        print(f'✔ Arquivo salvo: {nome_saida}')
    else:
        print(f'⚠ Nenhum arquivo válido encontrado na pasta: {pasta}')
```

```python
# Dicionário com os nomes a substituir
substituicoes_escola = {
    'ESCOLA MUNICIPAL DE TEMPO INTEGRAL NEGRO COSME': 'ESCOLA MUNICIPAL TEMPO INTEGRAL NEGRO COSME',
    'UEB ENS FUND PROF JOSE GONCALVES DO AMARAL RAPOSO': 'UEB PROF JOSE GONCALVES DO AMARAL RAPOSO',
    'UEB ENS FUND PROF MARIA THERESA CABRAL': 'UEB ENS FUND PROF MARIA THEREZA CABRAL'
}

substituicoes_ano = {
    'ENSINO FUNDAMENTAL DE 9 ANOS - 2º ANO': '2º ANO',
    'ENSINO FUNDAMENTAL DE 9 ANOS - 5º ANO': '5º ANO',
    'ENSINO FUNDAMENTAL DE 9 ANOS - 9º ANO': '9º ANO',
}
```

```python
# Lista com os nomes dos arquivos a tratar
# arquivos = [
#     'HABILIDADE_DESEMPENHO_ESCOLA_2_A_9_unificado.xlsx',
#     'HABILIDADE_DESEMPENHO_TURMA_1_A_9_unificado.xlsx'
# ]
arquivos = [
    'HABILIDADE_DESEMPENHO_TURMA_1_A_9_unificado.xlsx'
]

for arquivo in arquivos:
    df = pd.read_excel(arquivo)

    # Aplica as substituições na coluna 'Escola'
    df['Escola'] = df['Escola'].replace(substituicoes_escola)
    
    # Aplica as substituições na coluna 'Escola'
    df['Ano Escolar'] = df['Ano Escolar'].replace(substituicoes_ano)

    # Salva com o mesmo nome (sobrescreve)
    df.to_excel(arquivo, index=False)
    print(f'✔ Nomes padronizados no arquivo: {arquivo}')
```

```python
arquivos_com_prefixos = {
    'HABILIDADE_DESEMPENHO_ESCOLA_2_A_9_unificado.xlsx': 'Escola_',
    'HABILIDADE_DESEMPENHO_MUNICIPIO_1_A_9_unificado.xlsx': 'Municipio_',
}

for arquivo, prefixo in arquivos_com_prefixos.items():
    print(f'\n📁 Atualizando colunas do arquivo: {arquivo}')
    
    df = pd.read_excel(arquivo)

    # Evita repetir o prefixo em colunas que já o tenham ou preservar colunas específicas
    novas_colunas = {
        col: f"{prefixo}{col}" for col in df.columns if not col.startswith(prefixo)
    }

    df.rename(columns=novas_colunas, inplace=True)

    # Salva sobrescrevendo o próprio arquivo
    df.to_excel(arquivo, index=False)

    print(f'✅ Colunas renomeadas com prefixo "{prefixo}".')
    print(f'🔍 Novos nomes de colunas: {list(df.columns)[:5]}...')
```
